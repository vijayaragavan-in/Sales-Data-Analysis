## 8. Data Cleaning and Preprocessing

Data cleaning is an important step in the analysis process. It ensures that the dataset is consistent and suitable for analysis.

The following cleaning operations are performed:

1. Check for missing values.
2. Check for duplicate records.
3. Convert the order date into datetime format.
4. Handle missing categorical values.
5. Remove duplicate records.

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df["Region"] = df["Region"].fillna("Unknown")
df["Channel"] = df["Channel"].fillna("Unknown")
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df.isnull().sum()
```

## 9. Feature Engineering

Additional variables are created from the existing data to support deeper analysis.

The following variables are created:

- Profit Margin
- Month
- Year-Month

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["Profit_Margin"] = (df["Profit"] / df["Revenue"]) * 100
```

```python
df["Month"] = df["Order_Date"].dt.month_name()
```

```python
df.head()
```

## 10. Key Performance Indicators (KPIs)

Key Performance Indicators are used to measure the overall performance of the business.

The following KPIs are calculated:

- Total Revenue
- Total Cost
- Total Profit
- Profit Margin
- Total Orders
- Total Units Sold
- Average Order Value

```python


total_revenue = df["Revenue"].sum()
total_cost = df["Cost"].sum()
total_profit = df["Profit"].sum()
total_orders = len(df)
total_units = df["Quantity"].sum()
average_order_value = df["Revenue"].mean()
profit_margin = (total_profit / total_revenue) * 100

print("SALES PERFORMANCE KPIs")
print("-" * 40)
print(f"Total Revenue       : ₹{total_revenue:,.2f}")
print(f"Total Cost          : ₹{total_cost:,.2f}")
print(f"Total Profit        : ₹{total_profit:,.2f}")
print(f"Profit Margin       : {profit_margin:.2f}%")
print(f"Total Orders        : {total_orders:,}")
print(f"Total Units Sold    : {total_units:,}")
print(f"Average Order Value : ₹{average_order_value:,.2f}")
```

## 11. Product Analysis

Product-level analysis helps identify which products contribute most to revenue, profit, and sales volume.

```python

product_sales = df.groupby("Product")["Revenue"].sum().sort_values(ascending=False)

print(product_sales)
```

```python

product_profit = df.groupby("Product")["Profit"].sum().sort_values(ascending=False)

print(product_profit)
```

```python

product_units = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

print(product_units)
```

```python
plt.figure(figsize=(10, 6))

product_sales.plot(kind="bar")

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## 12. Regional Analysis

## 16. Correlation Analysis

Correlation analysis is used to understand relationships between numerical variables such as quantity, unit price, revenue, cost, and profit.

```python
numeric_columns = [
    "Quantity",
    "Unit_Price",
    "Discount",
    "Revenue",
    "Cost",
    "Profit",
    "Profit_Margin"
]

correlation = df[numeric_columns].corr()

correlation
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()
```

## 17. Top Performing Products
